```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df_superstore = pd.read_csv('/kaggle/input/superstore-sales-dataset/samplesuperstore.csv')

# Convert Order Date to datetime
df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])

# Set Order Date as index
df_superstore.set_index('Order Date', inplace=True)

# Resample Sales Weekly using sum aggregation
weekly_sales = df_superstore['Sales'].resample('W').sum()

# Plot the chart
weekly_sales.plot(figsize=(12, 6), title='Weekly Sales')

plt.xlabel('Week')
plt.ylabel('Total Sales')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Order Date is already the index
df_superstore.index = pd.to_datetime(df_superstore.index)

# Calculate daily sales
daily_sales = df_superstore['Sales'].resample('D').sum()

# Calculate 7-day rolling mean
rolling_avg = daily_sales.rolling(window=7).mean()

# Plot original daily sales and rolling average
plt.figure(figsize=(12, 6))

plt.plot(daily_sales, label='Daily Sales')
plt.plot(rolling_avg, label='7-Day Rolling Average')

plt.title('Daily Sales and 7-Day Rolling Average')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Check whether Order Date is already the index
if 'Order Date' in df_superstore.columns:
    df_superstore['Order Date'] = pd.to_datetime(df_superstore['Order Date'])
    df_superstore.set_index('Order Date', inplace=True)
else:
    df_superstore.index = pd.to_datetime(df_superstore.index)

# Calculate daily sales
daily_sales = df_superstore['Sales'].resample('D').sum()

# Calculate 7-day rolling mean
rolling_avg = daily_sales.rolling(window=7).mean()

# Plot original daily sales and rolling average
plt.figure(figsize=(12, 6))
plt.plot(daily_sales, label='Daily Sales')
plt.plot(rolling_avg, label='7-Day Rolling Average')

plt.title('Daily Sales and 7-Day Rolling Average')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

Task 3: Calculate Period-over-Period Daily Growth Using .Shift()

```python
import pandas as pd

# Calculate daily sales
daily_sales = df_superstore['Sales'].resample('D').sum()

# Create Lag-1 column using shift(1)
daily_sales = daily_sales.to_frame(name='Sales')

daily_sales['Sales_Prev_Day'] = daily_sales['Sales'].shift(1)

# Calculate daily percentage change in Sales
daily_sales['Sales_Pct_Change'] = daily_sales['Sales'].pct_change() * 100

# Display the result
print(daily_sales)
```

Task 2: Matplotlib 3D Surface Plot of Synthetic Sales-Profit Surface

```python
import numpy as np
import matplotlib.pyplot as plt

# Create data
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Create a meshgrid
X, Y = np.meshgrid(x, y)

# Define the surface
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create 3D figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot smooth 3D surface
surface = ax.plot_surface(
    X, Y, Z,
    cmap='viridis',
    edgecolor='none',
    antialiased=True
)

# Labels and title
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Smooth 3D Surface Plot')

# Add color bar
fig.colorbar(surface, ax=ax, shrink=0.6, label='Z Value')

plt.show()
```

```python
# Task 10: KPI Cards and Category Pie Chart

html.Div([
    html.Div(id='kpi-sales'),
    html.Div(id='kpi-profit'),
    html.Div(id='kpi-quantity')
]),

dcc.Graph(id='category-pie-chart')


@app.callback(
    Output('kpi-sales', 'children'),
    Output('kpi-profit', 'children'),
    Output('kpi-quantity', 'children'),
    Output('category-pie-chart', 'figure'),
    Input('region-dropdown', 'value')
)
def update_kpis_and_pie(selected_region):

    if selected_region == 'All':
        filtered_df = df_superstore
    else:
        filtered_df = df_superstore[
            df_superstore['Region'] == selected_region
        ]

    total_sales = filtered_df['Sales'].sum()
    total_profit = filtered_df['Profit'].sum()
    total_quantity = filtered_df['Quantity'].sum()

    category_sales = (
        filtered_df.groupby('Category')['Sales']
        .sum()
        .reset_index()
    )

    fig = px.pie(
        category_sales,
        names='Category',
        values='Sales',
        title='Sales by Category'
    )

    return (
        f"Total Sales: ${total_sales:,.2f}",
        f"Total Profit: ${total_profit:,.2f}",
        f"Total Quantity: {total_quantity:,}",
        fig
    )
```
